# BRCA Visium–Xenium Benchmark

This planned real-data benchmark will use the publicly available paired Xenium and Visium datasets from adjacent human breast cancer tissue sections described in [Joint imputation and deconvolution of gene expression across spatial transcriptomics platforms](https://pmc.ncbi.nlm.nih.gov/articles/PMC11870578/) (source dataset reference [19]). Appendix C.1 documents Xenium In Situ Sample 1, Replicate 1, Visium Spatial, FRP single-cell RNA sequencing, and the BRCA cell-type annotations; Appendix C.2 reports the dataset statistics.

Benchmark implementation and results will be added in a future update.
