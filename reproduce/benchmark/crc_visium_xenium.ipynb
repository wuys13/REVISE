# CRC Visium–Xenium Benchmark

This planned real-data benchmark will use the publicly available paired Xenium and Visium datasets from adjacent human colorectal cancer tissue sections described in [Joint imputation and deconvolution of gene expression across spatial transcriptomics platforms](https://pmc.ncbi.nlm.nih.gov/articles/PMC11870578/) (source dataset reference [32]). Appendix C.1 documents Visium CytAssist V2, Sample P2 CRC, Xenium In Situ, Sample P2 CRC, and the matching subset of aggregated Chromium Single Cell Flex data; Appendix C.2 reports the dataset statistics.

Benchmark implementation and results will be added in a future update.
